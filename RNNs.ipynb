```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, SimpleRNN, LSTM , GRU, Embedding, Input, Bidirectional
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
from tensorflow.keras.datasets import imdb

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train, maxlen=100)
X_test = pad_sequences(X_test, maxlen=100)
```

```python
X_train
```

```python
model = Sequential()

model.add(Input(shape=(100,)))
model.add(Embedding(input_dim=10000, output_dim=128, ))
model.add(LSTM(units=64))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history = model.fit(X_train,y_train, epochs=50, batch_size=32, validation_data=(X_test, y_test))
```

```python
acc,loss = model.evaluate(X_test,y_test)
```

```python
model1 = Sequential()

model1.add(Input(shape=(100,)))
model1.add(Embedding(input_dim=10000, output_dim=128, ))
model1.add(GRU(units=64))
model1.add(Dense(1, activation='sigmoid'))
```

```python
model1.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history1 = model1.fit(X_train,y_train, epochs=5, batch_size=32, validation_data=(X_test, y_test))
```

```python
model2 = Sequential()

model2.add(Input(shape=(100,)))
model2.add(Embedding(input_dim=10000, output_dim=128, ))
model2.add(Bidirectional(LSTM(units=64)))
model2.add(Dense(1, activation='sigmoid'))
```

```python
model2.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history2 = model2.fit(X_train,y_train, epochs=5, batch_size=32, validation_data=(X_test, y_test))
```
